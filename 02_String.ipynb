# Strings and Strings Methods

```python
name = "India is my country"
```

#### Print, Lower/Upper/Title, String-Length

```python
print(f"StringAsIs    = {name}");
print(f"Lower         = {name.lower()}");
print(f"Upper         = {name.upper()}");
print(f"Title         = {name.title()}");
print(f"StringLength  = {len(name)}")
```

### strip() / lstrip() / rstrip()

```python
s = "   hello world   "
print(s.strip())
```

### Replace

```python
str = "I Love Java"
print("Replace Java with python - ", str.replace("Java", "Python"))
```

### Split Delimeter

```python
str = "a,b,c"
print("Splitted String: ", str.split(","))
```

### .join(iterable) - Join list elements into a string.

```python
wordlist = ["I ","love "," India"]
str = "";
str = str.join(wordlist)
print("Joinned list:", str)
```

#### .find(sub) / .rfind(sub)

```python
str = "I love India"
print("Index of l from Left :    ", str.find("I"));
print("Index of I from Right:    ", str.rfind("I"));
print("Index of string no found: ", str.find("X"));
```

### .startswith() / .endswith()

```python
str = "I love India"
print("Does str starts with I : ", str.startswith("I"));
print("Does str ends with   I : ", str.endswith("I"));
```

### .count(sub)

```python
str = "I love India and love all Indians"

# case sensitive count
print("Count(I)    : ", str.count("I"));

#count even the occurance the Substrings
print("Count(love) : ", str.count("love"));
```

## String slicing [start:end:step]
 1. Start : Inclusive
 2. End   : Exclusive
 3. Step  : 
     - +ve: Skip chars
     - -ve: Skip chars AND rev

#### Basic Slicing And Omitting Start/End

```python
str = "Python Programming"

print("Substring: S:E :",str[0:6])   # Python
print("Substring: _:E :",str[:6])    # Python
print("Substring: S:E :",str[7:18])  # Programming
print("Substring: S:_ :",str[7:])    # Programming
```

#### Using Step

```python
str = "Python Programming"

print("Stepping: Every Char     :",str[::1])
print("Stepping: Every 2nd Char :",str[::2])
print("Stepping: Every 3nd Char :",str[::3])
```

#### Negative Indexing

```python
str = "Python-Programming"
print("Reverse String        : '" + str[::-1] + "'")
print("Reverse First 7 chars : '" + str[6::-1] + "'")
print("Reverse Last 11 chars : '" + str[:-12:-1] + "'")
```
